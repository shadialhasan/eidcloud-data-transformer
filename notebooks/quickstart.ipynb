```python
# [Cell 1] Setup Environment & Generate Sample Dataset
!apt-get install -y php-cli php-xml php-mbstring > /dev/null
!php -v

# Clone or link repository
!if [ ! -d 'eidcloud-data-transformer' ]; then git clone https://github.com/shadialhasan/eidcloud-data-transformer.git; fi
%cd eidcloud-data-transformer

# Create sample e-commerce customers CSV dataset
sample_csv = """id,full_name,email,age,status,balance,registered_at
1,Alice Al-Hasan,alice@eidcloud.sy,28,active,1450.50,2026-01-10 09:15:00
2,Bob Smith,bob@example.com,34,inactive,250.00,2026-02-14 14:30:00
3,Charlie Brown,charlie@eidcloud.sy,22,active,890.75,2026-03-01 11:45:00
4,Diana Prince,diana@themyscira.io,45,active,5200.00,2025-11-20 18:00:00
5,Evan Wright,evan@example.org,19,pending,0.00,2026-03-25 16:20:00
"""

with open('customers.csv', 'w') as f:
    f.write(sample_csv.strip())

print("Sample customers.csv generated successfully!")
```

```python
# [Cell 2] Convert CSV to JSON with Filtering
!php bin/eidcloud-data convert customers.csv \
  --to=json \
  --where="status=active and age>21" \
  --select="id,full_name,email,balance" \
  --sort="balance:desc" \
  --output=active_customers.json \
  --json

# Display generated JSON content
print("\n--- Generated JSON Content ---")
!cat active_customers.json
```

```python
# [Cell 3] Generate Production SQL with Automatic Type Inference
!php bin/eidcloud-data convert customers.csv \
  --to=sql \
  --table=customers \
  --where="age >= 21" \
  --sort="registered_at:desc" \
  --output=customers.sql

# Display generated SQL script (CREATE TABLE + INSERT statements)
print("\n--- Generated SQL Output ---")
!cat customers.sql
```
